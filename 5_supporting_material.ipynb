## FIR visualisation

```python
import json
import geopandas as gpd
import pandas as pd
import topojson

# ------------------------------------------------
# 1. Load the FIR TopoJSON
# ------------------------------------------------
with open(r"data/worldfirs.json") as f:
    topo = json.load(f)

gdf = (
    topojson.Topology(topo, object_name="data")
        .to_gdf()
        .set_crs(4326)
)

# ------------------------------------------------
# 2. Your reference list of "European" FIRs
# ------------------------------------------------
european_fir_list = [
    ("EGTT","London FIR"),("EDGG","Langen FIR"),("LFFF","Paris FIR"),("LECM","Madrid FIR"),
    ("EHAA","Amsterdam FIR"),("LECB","Barcelona FIR"),("GCCC","Canarias FIR"),("EDMM","Munchen FIR"),
    ("LPPC","Lisboa FIR"),("LIMM","Milano FIR"),("LIRR","Roma FIR"),
    ("EBBU","Brussels FIR"),("LSAS","Switzerland FIR"),("EPWW","Warszawa FIR"),("LGGG","Athinai FIR"),
    ("EDWW","Bremen FIR"),("EKDK","Copenhagen FIR"),("ESAA","Sweden FIR"),("EFIN","Helsinki FIR"),
    ("EISN","Shannon FIR"),("LOVV","Wien FIR"),("ENOR","Polaris FIR"),("LRBB","Bucarest FIR"),
    ("EGPX","Scottish FIR"),("LHCC","Budapest FIR"),("LKAA","Praha FIR"),("LFMM","Marseille FIR"),
    ("BIRD","Reykjavik FIR"),("LBSR","Sofia FIR"),("LFEE","Reims FIR"),
    ("EYVL","Vilnius FIR"),("EVRR","Riga FIR"),("LDZO","Zagreb FIR"),
    ("LFBB","Bordeaux FIR"),("LMMM","Malta FIR"),("LYBA","Beograd FIR"),("EETT","Tallinn FIR"),
    ("LIBB","Brindisi FIR"),("LFRR","Brest FIR"),("LUUU","Chisinau FIR"),("LPPO","Santa Maria FIR"),
    ("LZBB","Bratislava FIR"),("LWSS","Skopje FIR"),("LAAA","Tirana FIR"),("LJLA","Ljubljana FIR"),
    ("LQSB","Sarajevo FIR"),("EGGX", "Shanwick FIR")
]

european_designators = [des for des, name in european_fir_list]
# ------------------------------------------------
# 3. Filter only FIRs, then tag European/Non-European
# ------------------------------------------------
firs = gdf[gdf["type"].str.upper() == "FIR"].copy()

# Keep only the ones in our European list
european_firs = firs[firs["designator"].isin(european_designators)]

# ------------------------------------------------
# 4. Save to CSV or GeoJSON
# ------------------------------------------------
european_firs.to_file("data/european_firs.geojson", driver="GeoJSON")

firs["is_european"] = firs["designator"].isin(european_designators["designator"])
firs["region_class"] = firs["is_european"].map({True: "European", False: "Non-European"})

# Add a unique ID
firs["id"] = firs["designator"] + "_" + firs["name"].str.replace(" ", "_")

# ------------------------------------------------
# 4. Export for Flourish
# ------------------------------------------------
firs.to_file("data/worldfirs.geojson", driver="GeoJSON")
```
